```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python
# Load a pretrained YOLOv8 detection model
model = YOLO('yolov8n.pt')  # Replace 'yolov8n.pt' with your desired model
```

```python
from google.colab import files

# Upload your image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]  # Get the uploaded file path
```

```python
# Make predictions
results = model(image_path)

# Print the results
print(results)
```

```python
# Load the original image
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert to RGB for Matplotlib

# Visualize predictions
annotated_image = results[0].plot()  # Use YOLO's built-in method to annotate the image

# Display the image
plt.figure(figsize=(10, 10))
plt.imshow(annotated_image)
plt.axis('off')
plt.show()
```

```python
# Annotate the image
annotated_image = results[0].plot()

# Convert BGR to RGB
annotated_image_rgb = cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB)

# Display the annotated image
plt.figure(figsize=(10, 10))
plt.imshow(annotated_image_rgb)
plt.axis('off')
plt.show()
```

```python
# Annotate the image
annotated_image = results[0].plot()

# Save the annotated image
output_path = "finalApple.jpg"  # Choose your file name
cv2.imwrite(output_path, annotated_image)

# Download the saved image to your PC
from google.colab import files
files.download(output_path)
```
