###Visualizing box plot for identifying outliers in numerical data
Visualize outliers if any by creating boxplots for numeric features
We will be creating subplot for all the numeric features.


When we create multiple plots inside one figure that is what we refer to as the subplot.

It needs to know how many rows and columns to use to create different plots.

plt.subplot(rows, cols, position) here position indicates out of all the rows & columns, which position to plot in.

For example, plt.subplot(2, 3, 2) means that create a plot on the second postion of 6 available positions.






```python
# plot a box plot using seaborn library
plt.style.use('default')
plt.figure(figsize=(25,12))
plt.suptitle(' Visualization of various numerical aspect with respect to star type' , fontsize='20') # writes the title of subplot
#  sns.boxplot(x=star_df['Star type'],y=star_df.iloc[:,0])   # plots a boxplot with x axis as star type and y axis as temperature
for index in range(4):
    plt.subplot(2,2,index+1)# matplotlib's subplot feature it takes argumenets as (no. of rows,no. of column, position of plt to drawn)
    sns.boxplot(x=star_df['Star type'],y=star_df.iloc[:,index])
    plt.title(star_df.columns[index])
    plt.ylabel('')
plt.savefig(base_dir + 'boxplot_star_type.jpg')
plt.show()
```

```python
# visualization various numerical aspect of star with respect to star color
plt.figure(figsize=(25,12))
plt.suptitle(' Visualization of various numerical aspect with respect to star color' , fontsize='20') # writes the title of subplot
#  sns.boxplot(x=star_df['Star type'],y=star_df.iloc[:,0])   # plots a boxplot with x axis as star type and y axis as temperature
for index in range(4):
    plt.subplot(2,2,index+1)# matplotlib's subplot feature it takes argumenets as (no. of rows,no. of column, position of plt to drawn)
    sns.boxplot(x=star_df['Star color'],y=star_df.iloc[:,index])
    plt.title(star_df.columns[index])
    plt.ylabel('')
plt.savefig(base_dir + 'boxplot_star_color.jpg')
plt.show()
```

###Creating  Line plots  
1. line plots are use to show distribution in the data

```python
def line_plotter(star_df,colors,i):
 plt.subplot(4,1,i+1)
 plt.plot(star_df.iloc[:,i],color=colors[i])
 plt.title(star_df.columns[i],color='red')
```

```python
plt.figure(figsize=(20,8))
plt.style.use('dark_background')
colors = ['yellow','orange','blue','white']
plt.suptitle('Distribution in numerical feature in dataset')
for index in range(4):
  line_plotter(star_df,colors,index)
plt.tight_layout()
plt.savefig(base_dir + "lineplot_dist.jpg")
plt.show()
```

### H-R diagram(Hertzsprung-Russell diagram)


```python
#H-R diagram is temp vs abs mag
star_type = star_df['Star type'].values;# getiing values of star types
temperature = star_df['Temperature (K)'].values;
abs_mag = star_df['Absolute magnitude(Mv)'].values;
# creating a dictionary with key(star_type) and values ( label,color,size of marker , marker type)
star_type_properties = {
     0: {'label': 'Brown Dwarf', 'color': 'brown', 'size': 30, 'marker': '.'},
    1: {'label': 'Red Dwarf', 'color': 'red', 'size': 35, 'marker': '.'},
    2: {'label': 'White Dwarf', 'color': 'white', 'size': 40, 'marker': '.'},
    3: {'label': 'Main Sequence', 'color': 'cyan', 'size': 30, 'marker': 'o'},
    4: {'label': 'Supergiants', 'color': 'orange', 'size': 100, 'marker': 'o'},
    5: {'label': 'Hypergiants', 'color': 'maroon', 'size': 150, 'marker': 'o'}
}
#key:values
axes = []
labels = set()
plt.figure(figsize = (10,6))
for i in range(len(star_type)):
    properties = star_type_properties[star_type[i]]
    if properties['label']  not in labels:
      ax = plt.scatter(temperature[i],abs_mag[i],
                       s= properties['size'],
                       c= properties['color'],
                       marker =properties['marker'],
                       label= properties['label'])
      axes.append(ax)
      labels.add(properties['label'])

    else:
       plt.scatter(temperature[i],abs_mag[i],
                       s= properties['size'],
                       c= properties['color'],
                       marker =properties['marker'] )

# adding data of sun
ax_sun = plt.scatter(5778, 4.83, s=75, c="yellow", marker='o', label="Sun")
axes.append(ax_sun)
labels.add('Sun')

plt.title(f" H-R diagram for {len(star_type)}",color = "blue" , weight = "bold",size = 17)
plt.ylabel(" Absolute Magnitude ", color = "tab:pink",size = 15)
plt.xlabel(" Temperature ", color = "tab:pink",size = 15)


plt.legend(handles = axes)


plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.show()
plt.savefig(base_dir + "H-R diagram.jpg")
```
