```python
import pandas as pd
df = pd.read_csv('Social_Network_Ads.csv')
df.head()
```

```python
dataset = df.drop(['User ID','Gender'],axis = 1)

```

```python
# Purchased as categorical dataset
dataset['Purchased'] = dataset['Purchased'].astype('category')
X = dataset.drop('Purchased',axis = 1)
y = dataset['Purchased']
```

```python
# training testing split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train , y_test = train_test_split(X,y,test_size=0.3,random_state=0)

```

```python
# scaling values of dataset
from sklearn.preprocessing  import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

```

```python
from sklearn.svm import SVC

model = SVC(kernel = 'linear')
model.fit(X_train,y_train)
y_pred = model.predict(X_test)

# evalution of model

from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred)
print(cm)
```

```python
import seaborn as sns

sns.heatmap(cm,annot=True)
```

```python
# Step 1: Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Step 2: Create a synthetic dataset
# Create a dataset with 3 clusters
X, y = make_blobs(n_samples=300, centers=3, cluster_std=0.5, random_state=0)

# Step 3: Visualize the synthetic dataset
plt.scatter(X[:, 0], X[:, 1], s=30)
plt.title('Synthetic Dataset')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()

# Step 4: Initialize and fit the K-Means model
k = 3  # Number of clusters
kmeans = KMeans(n_clusters=k)
kmeans.fit(X)

# Step 5: Get cluster centers and labels
centers = kmeans.cluster_centers_
labels = kmeans.labels_

# Step 6: Visualize the clustering results
plt.scatter(X[:, 0], X[:, 1], c=labels, s=30, cmap='viridis')
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X', label='Centroids')
plt.title('K-Means Clustering Results')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```
